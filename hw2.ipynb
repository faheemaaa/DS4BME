## 1. Sorting Hat

`sorting_hat` asks two yes/no questions (using `input`) and uses `if`/`elif`/`else`
logic to sort the caller into Gryffindor or Slytherin. The chosen house is returned.

```python
def sorting_hat():
    """
    Ask two Y/N questions and sort the caller into a Hogwarts house
    (Gryffindor or Slytherin) using if/elif/else logic.

    Returns
    -------
    str
        The name of the assigned Hogwarts house.
    """
    # --- Question 1: bravery ---
    # input() pauses execution and waits for the user to type a response.
    # .strip() removes leading/trailing whitespace (e.g. accidental spaces).
    # .upper() normalizes the answer so "y", "Y", " y " all count the same.
    q1 = input("Do you value bravery above all else? (Y/N): ").strip().upper()

    # --- Question 2: ambition ---
    # Same pattern as above, just a different question.
    q2 = input("Would you do whatever it takes to achieve your ambitions? (Y/N): ").strip().upper()

    # --- Sorting logic ---
    # We check the combination of both answers using if/elif/else.
    # Case 1: brave (Y) but not especially ambitious (N) -> Gryffindor.
    if q1 == "Y" and q2 == "N":
        house = "Gryffindor"
    # Case 2: not particularly brave (N) but very ambitious (Y) -> Slytherin.
    elif q1 == "N" and q2 == "Y":
        house = "Slytherin"
    # Case 3: both brave AND ambitious (Y, Y) -> ambition wins, Slytherin.
    elif q1 == "Y" and q2 == "Y":
        house = "Slytherin"
    # Case 4 (the "else" catch-all): neither bold nor ambitious (N, N),
    # or any unrecognized/invalid input -> default to Gryffindor.
    else:
        house = "Gryffindor"

    # Announce the result to whoever is running the cell...
    print(f"The Sorting Hat has placed you in... {house}!")
    # ...and also return it, since the assignment requires the house
    # to be the function's return value (not just printed).
    return house
```

```python
# Call the function -- this will prompt you (twice) for Y/N input.
my_house = sorting_hat()

# Displaying the variable on the last line shows the returned value
# as the cell's output (in addition to the print() inside the function).
my_house
```

## 2. `mydata` Class

`mydata` stores a list of numbers and computes the mean and (population) standard
deviation using only base Python (no `numpy`/`statistics`).

```python
class mydata:
    """
    Store a list of numbers and compute descriptive statistics
    using only base Python.
    """

    def __init__(self, numbers):
        # The constructor runs automatically when we do mydata([...]).
        # We copy the input into a list and save it on the instance
        # (self.numbers) so mean() and std() can use it later.
        self.numbers = list(numbers)

    def mean(self):
        """Return the arithmetic mean of the stored numbers."""
        # mean = (sum of all values) / (how many values there are).
        # sum() and len() are both base Python built-ins -- no numpy needed.
        return sum(self.numbers) / len(self.numbers)

    def std(self):
        """Return the (population) standard deviation of the stored numbers."""
        # Step 1: get the mean, since std is measured as spread around it.
        mu = self.mean()

        # Step 2: variance = average of the squared differences from the mean.
        # (x - mu) ** 2 measures how far each value x is from the mean,
        # squared so negative and positive differences don't cancel out.
        # This is a generator expression fed straight into sum().
        variance = sum((x - mu) ** 2 for x in self.numbers) / len(self.numbers)

        # Step 3: standard deviation is the square root of the variance
        # (** 0.5 is the base-Python way to take a square root).
        return variance ** 0.5
```

Try it out:

```python
# Create an instance of mydata holding the numbers [1, 2, 7, 5].
x = mydata([1, 2, 7, 5])

# Call the mean() and std() methods on that instance and print the results.
print("mean:", x.mean())
print("std:", x.std())
```

## 3. `lag` Function

`lag` takes a numpy vector and an integer `n > 0`, and returns a new numpy array
shifted forward by `n` positions: the last `n` values are dropped and `n`
`numpy.nan` values are prepended. This uses plain numpy indexing/slicing only
(no `numpy.roll`, lag/lead helpers, etc.).

```python
import numpy

def lag(a, n):
    """
    Lag a numpy vector by n positions.

    Parameters
    ----------
    a : numpy.ndarray
        Input vector.
    n : int
        Number of positions to lag by (n > 0).

    Returns
    -------
    numpy.ndarray
        A new array of the same length as `a`, with the first `n` entries
        set to numpy.nan and the remaining entries equal to `a` shifted
        forward by `n` (i.e. the last `n` values of `a` are dropped).
    """
    # Make sure we're working with a numpy array of floats (not ints),
    # because numpy.nan only exists as a float -- an int array can't hold it.
    a = numpy.asarray(a, dtype=float)

    # Create an empty array the same length/shape as `a` to hold the result.
    # "empty" just means uninitialized memory -- we're about to fill it in
    # completely below, so we don't need numpy to waste time zeroing it out.
    result = numpy.empty(a.shape, dtype=float)

    # First `n` entries of the result become numpy.nan (missing values),
    # since there's no earlier data to shift into those slots.
    result[:n] = numpy.nan

    # The remaining entries (from index n onward) become the ORIGINAL
    # array's values, but only up to len(a) - n of them -- i.e. we drop
    # the last n values of `a` because they get pushed off the end.
    # Example: a = [1, 6, 8, 4], n = 1
    #   result[:1]  = [nan]              <- 1 missing value at the front
    #   result[1:]  = a[:3] = [1, 6, 8]  <- first 3 of the original 4 values
    #   final result = [nan, 1, 6, 8]
    result[n:] = a[: len(a) - n]

    return result
```

Try it out:

```python
# The example vector from the assignment.
a = numpy.array([1, 6, 8, 4])

# Lagging by 1: drop the last value (4), prepend one nan.
# Expected: [nan, 1, 6, 8]
print(lag(a, 1))

# Lagging by 2: drop the last two values (8, 4), prepend two nans.
# Expected: [nan, nan, 1, 6]
print(lag(a, 2))
```

## 4. `lagGenerator` Function

`lagGenerator` takes an integer `n` and returns a *function* (built with `lambda`)
that lags any vector passed to it by `n`, built on top of `lag` from Problem 3.

```python
def lagGenerator(n):
    """
    Return a function that lags its input vector by n positions.

    Parameters
    ----------
    n : int
        The lag amount the generated function will apply.

    Returns
    -------
    function
        A function f(a) that returns lag(a, n).
    """
    # `lagGenerator` is a "function factory": instead of returning a value,
    # it returns a brand-new FUNCTION that remembers the value of `n` it
    # was created with (this is called a closure).
    #
    # `lambda a: lag(a, n)` is shorthand for:
    #     def temp_function(a):
    #         return lag(a, n)
    #     return temp_function
    #
    # Every time you call lagGenerator(some_n), you get back a fresh
    # one-argument function that always lags whatever vector you give it
    # by that same fixed `some_n`.
    return lambda a: lag(a, n)
```

Try it out:

```python
# lagGenerator(2) hands back a ready-to-use function that always lags by 2.
lag_by_2 = lagGenerator(2)
print(lag_by_2(a))  # same result as lag(a, 2): [nan, nan, 1, 6]

# We can make as many of these specialized functions as we want.
# lagGenerator(1) gives us one that always lags by 1.
lag_by_1 = lagGenerator(1)
print(lag_by_1(a))  # same result as lag(a, 1): [nan, 1, 6, 8]
```
