```python
class mydata:
    def __init__(self, numbers):
        self.data = list(numbers)

    def mean(self):
        if not self.data:
            return 0
        return sum(self.data) / len(self.data)

    def std(self):
        n = len(self.data)
        if n == 0:
            return 0
        mu = self.mean()
        # Population standard deviation using only base Python: sqrt(sum((x - mu)^2) / n)
        variance = sum((x - mu) ** 2 for x in self.data) / n
        return variance ** 0.5

# Test Problem 2:
x = mydata([1, 2, 7, 5])
print("Data:", x.data)
print("Mean:", x.mean())
print("Standard Deviation:", x.std())
```

```python
import numpy as np

def lag(a, k):
    """
    Lags a 1D numpy array `a` by an integer `k` > 0 using numpy indexing.
    Replaces the first `k` positions with np.nan and slices the rest.
    """
    if k <= 0:
        return a.astype(float).copy()
    
    # Pre-allocate array of np.nan with the same length as a
    result = np.full(len(a), np.nan, dtype=float)
    
    if k < len(a):
        # Place elements from the start up to len(a) - k into indices starting at k
        result[k:] = a[:-k]
        
    return result

# Test Problem 3:
a = np.array([1, 6, 8, 4])
print("lag(a, 1):", lag(a, 1))
print("lag(a, 2):", lag(a, 2))
```

```python
def lagGenerator(k):
    """
    Returns a function that lags any given vector by `k` positions.
    """
    return lambda arr: lag(arr, k)

# Test Problem 4:
lag2 = lagGenerator(2)
print("lagGenerator(2)(a):", lag2(a))
```
